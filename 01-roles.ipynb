```python
 O BASTIDOR DA MÁGICA

# Trazendo a ferramenta mágica que esconde nossas senhas. 
# (Ninguém quer deixar a senha do cartão de bobeira na internet, né?)
from dotenv import load_dotenv

# Importando o 'Groq', que é o motor super rápido da nossa IA.
# (Pensa nele como se fosse o seu console de videogame: o PS5 ou o PC Gamer)
from groq import Groq

# O 'os' (Operating System) é o tradutor que conversa direto com o cérebro do seu computador.
import os

# Aqui a gente roda o comando: "Ei computador, abra o baú secreto e leia minhas senhas!"
load_dotenv()

# Ligando o controle! A variável 'client' é o nosso controle remoto para pilotar a IA.
client = Groq(
    # Para o servidor da IA deixar a gente entrar, precisamos do nosso "Passe VIP". 
    # O comando abaixo vai lá buscar a nossa chave secreta de forma invisível, 
    # garantindo que nenhum espião (ou colega do lado) copie a sua senha.
    api_key=os.environ.get("GROQ_API"), 
)
```

### Compreendendo os Roles

🤖 Aula 01: Quem manda aqui? Entendendo os Roles (Papéis)

Para a IA não ficar perdida, a gente precisa dividir quem é quem na conversa. Imagine que você está criando um personagem em um jogo:

System: É a "ficha do personagem". Aqui você define a personalidade, os poderes e as regras que ele deve seguir (Ex: "Você é um pirata que só fala rimando").

User: É você, o jogador. É quem dá as ordens e faz as perguntas.

Assistant: É o robô sendo o personagem que você criou.


ENTENDENDO O CODIGO (client.chat.completions.create): 
🍔 A Analogia do Delivery: chat_completion = client.chat.completions.create
client 👉 O seu celular. Lembra que lá em cima a gente configurou o client com a nossa senha secreta (API Key)? Então, ele é o aparelho autorizado a fazer o pedido.

.chat 👉 O aplicativo. Você abriu o app de mensagens (ou o iFood) no seu celular.

.completions 👉 A tela de digitar. É o espaço onde você vai colocar o seu pedido (o prompt que você escreveu). A palavra "completion" significa "completar", porque no fundo, a IA é apenas um robô muito inteligente tentando completar a frase que você começou.

.create(...) 👉 O botão de "ENVIAR". É aqui que não tem mais volta! Você apertou o botão e a mensagem foi enviada lá para os servidores gigantes da Groq.

chat_completion = 👉 A mochila do entregador. Quando a resposta da IA voltar da internet, ela precisa ser guardada em algum lugar, certo? O sinal de igual (=) pega a resposta gigante que a IA gerou e joga dentro dessa "caixa" (variável) chamada chat_completion.

```python
# Aqui nós chamamos a "mágica" para acontecer. O 'client' é o nosso controle de videogame.
chat_completion = client.chat.completions.create(
    # A lista 'messages' é o histórico do chat. Por enquanto, tem só uma mensagem.
    messages=[
        {
            "role": "user", # O papel de quem está falando (você, o jogador)
            "content": "Explica o que é uma LLM como se eu fosse um gamer viciado em Minecraft.",
        }
    ],
    # Aqui a gente escolhe qual "cérebro" a IA vai usar.
    model="openai/gpt-oss-120b", 
)

# Imprimindo a resposta na tela. A IA fica guardada dentro de 'choices'.
print("🤖 Resposta Padrão:")
print(chat_completion.choices[0].message.content)# Aqui nós chamamos a "mágica" para acontecer. O 'client' é o nosso controle de videogame.
chat_completion = client.chat.completions.create(
    # A lista 'messages' é o histórico do chat. Por enquanto, tem só uma mensagem.
    messages=[
        {
            "role": "user", # O papel de quem está falando (você, o jogador)
            "content": "Explica o que é uma LLM como se eu fosse um gamer viciado em Minecraft.",
        }
    ],
    # Aqui a gente escolhe qual "cérebro" a IA vai usar.
    model="openai/gpt-oss-120b", 
)

# Imprimindo a resposta na tela. A IA fica guardada dentro de 'choices'.
print("🤖 Resposta Padrão:")
print(chat_completion.choices[0].message.content)
```

### Definindo um comportamento para modelo

```python
# Agora vamos instalar um 'mod' na IA para mudar a personalidade dela!
chat_completion = client.chat.completions.create(
    messages=[
        # O PRIMEIRO passo é sempre o 'system' (a regra do jogo)
        {
            "role": "system", 
            "content": "Você é um mestre de RPG de fantasia medieval muito dramático. Use termos como 'nobre viajante', 'pergaminhos' e 'magia'."
        },
        # O SEGUNDO passo é o 'user' fazendo a pergunta dentro desse contexto
        {
            "role": "user", 
            "content": "Como eu faço para estudar para a prova de biologia sem dormir no livro?"
        }
    ],
    model="openai/gpt-oss-120b",
)

print("🧙 Resposta do Mestre de RPG:")
print(chat_completion.choices[0].message.content)
```

 Memória e Contexto (Assistant)

```python
# A IA tem "amnésia". Ela não lembra do que você falou há 5 minutos.
# Para ela lembrar, você precisa enviar o histórico da conversa TODA vez!
chat_completion = client.chat.completions.create(
    messages=[
        # 1. Regra geral
        {
            "role": "system",
            "content": "Você é um personal trainer de e-sports focado em levar o usuário ao nível Pro."
        },
        # 2. Algo que você já perguntou no passado
        {
            "role": "user",
            "content": "Minha mira no Valorant está horrível hoje, o que eu faço?"
        },
        # 3. A resposta que a PRÓPRIA IA ('assistant') já te deu antes
        {
            "role": "assistant",
            "content": "Soldado, você precisa de 15 minutos de Aim Lab e parar de beber energético quente! O foco é total!"
        },
        # 4. A sua NOVA pergunta (agora ela tem todo o contexto para responder)
        {
            "role": "user",
            "content": "E se eu treinar só 5 minutos?"
        }
    ],
    model="openai/gpt-oss-120b",
)

print("🎮 Resposta com Contexto (Memória):")
print(chat_completion.choices[0].message.content)
```

### Usando System Prompt Refinados

```python
# Quanto mais detalhes você colocar no 'system', melhor e mais específica a IA fica.
# É como dar um roteiro de Hollywood na mão dela.
chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": (
                "Você é uma avó muito fofa, mestre em tecnologia, que explica coisas complexas "
                "usando analogias de cozinha. Sempre chame o usuário de 'meu netinho' ou 'minha netinha'. "
                "Ao explicar, dê uma receita de 3 passos para o sucesso e termine a frase dizendo para ir comer uma fruta."
            ),
        },
        {
            "role": "user",
            "content": "Vovó, por que meu código Python está dando erro de indentação?"
        }
    ],
    model="openai/gpt-oss-120b",
)

print("👵 Resposta da Vovó Tech:")
print(chat_completion.choices[0].message.content)
```
