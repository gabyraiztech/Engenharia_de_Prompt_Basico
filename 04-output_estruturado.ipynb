## 📖 O Conceito

Controlar o formato da saída é como dar ao escriba do reino um **template de documento**.

Sem template → ele escreve um rolo de pergaminho confuso.
Com template → ele preenche campos específicos, na ordem certa.

Você passa o template no `system` prompt e a IA segue o molde.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Ficha de Personagem em Tabela

```python
system_ficha = """
Você é o sistema de fichas do jogo. SEMPRE responda no formato de tabela:

| Atributo | Valor |
|---|---|
| Nome | ... |
| Classe | ... |
| Nível | ... |
| HP | .../... |
| Força | ... |
| Agilidade | ... |
| Magia | ... |
"""

ficha = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_ficha},
        {"role": "user",   "content": "Crie a ficha de um arqueiro élfico nível 7 especializado em magia das sombras."}
    ],
    temperature=0.4
)
print(ficha.choices[0].message.content)
```

## 🔍 Demonstração 2 — Card de Item Mágico

```python
system_card = """
Você cria cards de itens mágicos no formato:

⚔️ ITEM: [nome]
✨ TIPO: [arma/armadura/poção/artefato]
🔮 PODER: [efeito em 1 frase]
⚠️ MALDIÇÃO: [efeito negativo, se houver]
💰 VALOR: [em moedas de ouro]
"""

card = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_card},
        {"role": "user",   "content": "Crie um card para uma espada que absorve a vida dos inimigos."}
    ],
    temperature=0.6
)
print(card.choices[0].message.content)
```

## 🔍 Demonstração 3 — Log de Missão em JSON

```python
import json

log_prompt = """
Gere um log de missão em JSON com as chaves:
- 'missao': nome da missão
- 'status': 'ativa', 'completa' ou 'falhou'
- 'objetivos': lista com 3 objetivos
- 'recompensa': dicionário com 'ouro' e 'item'
"""

log = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": log_prompt},
        {"role": "user",   "content": "Gere o log da missão: recuperar o amuleto roubado do castelo Sombraverde."}
    ],
    temperature=0.3
)
print("=== JSON BRUTO ===")
print(log.choices[0].message.content)
```

## 🔍 Demonstração 4 — Checklist de Dungeon

```python
checklist = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você cria checklists no formato:\n☐ [tarefa]\n(sem explicações extras)"},
        {"role": "user",   "content": "Crie um checklist de 6 itens para preparar um herói antes de entrar numa dungeon."}
    ],
    max_completion_tokens=200,
    temperature=0.4
)
print(checklist.choices[0].message.content)
```

## 🎯 Seu Turno

Crie um card de item mágico para um equipamento do seu jogo.

Mude o `system` para o formato que você quiser!

```python
# SEU TURNO — defina o formato e o item
MEU_FORMATO = """
Você cria cards de itens no formato:
🎒 NOME: ???
🔥 PODER: ???
???  # Adicione mais campos que você quiser!
"""

MEU_ITEM = "???"  # Ex: 'Um chapéu que faz o usuário invisível mas o deixa com soluço'

resposta = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": MEU_FORMATO},
        {"role": "user",   "content": f"Crie um card para: {MEU_ITEM}"}
    ],
    temperature=0.6
)
print(resposta.choices[0].message.content)
```

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Output estruturado | Saída em formato definido (tabela, JSON, lista) | Ficha de personagem preenchida corretamente |
| Template no system | Molde passado na instrução do sistema | O modelo de documento do escriba |
| JSON | Formato de dados chave-valor usado em programação | Banco de dados do jogo (inventário, quests) |
| Card | Formato visual compacto de informação | Card de item do baralho do RPG |
