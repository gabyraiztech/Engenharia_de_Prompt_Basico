> ❓ **O log de aventura do seu herói tem 10 sessões de jogo e está enorme. Como você cria um resumo épico que caiba em uma página?**

Hoje você aprende **sumarização** — condensar textos longos sem perder as informações importantes, como o escriba do reino que transforma 50 pergaminhos em um relatório de uma página.

## 🗺️ O Mapa desta Aula

| # | O que vamos fazer |
|---|---|
| 1 | Resumo básico de log de aventura |
| 2 | Resumo em bullet points |
| 3 | Resumo em diferentes estilos (técnico vs. épico) |
| 4 | Resumo em JSON para o sistema do jogo |

## 📖 O Conceito

**Sumarização** é pedir à IA que **reduza** um texto mantendo as informações mais importantes.

O segredo está no **system prompt**: você define o escriba — qual estilo ele usa, qual tamanho deve ter o resumo, o que priorizar.

| Escriba | Estilo | Usa quando... |
|---|---|---|
| Narrador épico | Dramático, heróico | Intro da sessão seguinte |
| Analista tático | Bullet points secos | Revisar o que aconteceu |
| Sistema do jogo | JSON estruturado | Salvar estado no banco de dados |

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração — Log de Aventura para Resumos

```python
log_aventura = """
Sessão 7: O grupo chegou à cidade de Khal'Maren. Aldric comprou uma espada encantada por 200 moedas.
Lyra descobriu que o estalajadeiro era um espião do Império. Após uma briga na taverna, o grupo fugiu
pelos telhados. Zara foi ferida por uma flecha envenenada — HP caiu para 12. Encontraram um mapa
rasgado com pistas para a Torre dos Ecos. Lyra decodificou a primeira metade do mapa.
O grupo dormiu num celeiro a 3 km da cidade. Sessão encerrada com o grupo em alerta máximo.
"""

# Resumo básico — 1/3 do tamanho original
resumo_basico = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é o Escriba do Reino. Crie um resumo com cerca de 1/3 do tamanho original, preservando eventos-chave, itens adquiridos e estado do grupo."},
        {"role": "user",   "content": log_aventura}
    ],
    temperature=0.2
)
print("=== RESUMO BÁSICO ===")
print(resumo_basico.choices[0].message.content)
```

```python
# Resumo em bullet points — formato de briefing tático
bullets = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um analista tático. Resuma em bullet points:\n- Eventos importantes\n- Itens adquiridos\n- Estado do grupo\n- Próximos objetivos conhecidos"},
        {"role": "user",   "content": log_aventura}
    ],
    temperature=0.2
)
print("\n=== BULLET POINTS ===")
print(bullets.choices[0].message.content)
```

```python
# Resumo épico — intro dramática para a próxima sessão
epico = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um narrador épico de RPG. Crie um resumo dramático de 3 frases para abrir a próxima sessão. Use linguagem heróica e poética."},
        {"role": "user",   "content": log_aventura}
    ],
    temperature=0.8
)
print("\n=== ABERTURA ÉPICA DA PRÓXIMA SESSÃO ===")
print(epico.choices[0].message.content)
```

```python
import json

# Resumo em JSON — para salvar o estado do jogo
json_state = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": 'Resuma o log em JSON com as chaves: "sessao", "eventos" (lista), "itens_adquiridos" (lista), "estado_grupo" (dicionário com HP e status), "proximos_objetivos" (lista).'},
        {"role": "user",   "content": log_aventura}
    ],
    temperature=0.2
)
print("\n=== JSON DO ESTADO ===")
print(json_state.choices[0].message.content)
```

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |

| Sumarização | Condensar texto preservando o essencial | O escriba resumindo o pergaminho longo |
