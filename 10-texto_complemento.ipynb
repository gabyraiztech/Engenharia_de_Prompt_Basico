## 📖 O Conceito

**Text completion** é passar um texto incompleto e pedir à IA para continuar.

O segredo: o contexto que você passa no início **molda todo o estilo** que a IA vai usar. Se você escreve em prosa medieval poética, ela continua em prosa medieval poética.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Continuando uma Lenda

```python
lenda_inicio = """
Conta a lenda que, nos primeiros dias do mundo, quando os deuses ainda caminhavam entre os mortais,
uma forja foi erguida no coração de uma montanha de obsidiana. Lá, o ferreiro divino Ignarok
moldou sete espadas — cada uma carregando um fragmento da alma do mundo.
A primeira espada foi dada aos guerreiros do norte. A segunda, aos guardiões do mar.
Mas a sétima espada...
"""

continuacao = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um bardo mestre. Continue a lenda no mesmo tom épico e poético. Mais 2 parágrafos."},
        {"role": "user",   "content": lenda_inicio}
    ],
    temperature=0.8
)
print("=== LENDA COMPLETA ===")
print(lenda_inicio)
print(continuacao.choices[0].message.content)
```

## 🔍 Demonstração 2 — Expandindo a Descrição de uma Cidade

```python
cidade_inicio = "Valdrak é uma cidade portuária no extremo sul do continente, famosa por seus mercadores"

# Expansão simples
exp1 = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é o narrador do guia de viagens deste mundo. Expanda a descrição em 1 parágrafo com arquitetura, clima e segredos."},
        {"role": "user",   "content": cidade_inicio}
    ],
    temperature=0.7
)
print("=== CIDADE EXPANDIDA ===")
print(exp1.choices[0].message.content)
```

## 🔍 Demonstração 3 — Lore de um Item

```python
item_prompt = """
Crie a lore completa do item abaixo:

--- ITEM ---
Nome: Amuleto das Três Luas
Tipo: Artefato mágico
Poder: Permite ver no escuro e falar com espíritos
--- FIM ---

Inclua: origem, quem criou, por que foi criado, onde foi perdido e quem o procura agora.
"""

lore = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": item_prompt}],
    temperature=0.75
)
print(lore.choices[0].message.content)
```

## 🎯 Seu Turno

Escreva o início de qualquer texto do seu mundo — lenda, descrição, lore — e deixe a IA completar.

```python
# SEU TURNO — escreva o início e a IA continua
MEU_INICIO = """
???  # Escreva as primeiras linhas da sua lenda, cidade ou item
"""

INSTRUCAO = "???"  # Ex: 'Continue em 2 parágrafos no mesmo tom', 'Expanda com 3 segredos'

r = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": f"Você é um bardo e escriba do mundo. {INSTRUCAO}"},
        {"role": "user",   "content": MEU_INICIO}
    ],
    temperature=0.8
)
print(r.choices[0].message.content)
```

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Text completion | Continuar um texto a partir de um início | O bardo improvisando o segundo verso |
| Expansão | Desenvolver um tópico curto em algo maior | A descrição de uma cidade em 1 linha → 1 página |
| Lore | História de fundo de um elemento do mundo | A origem e segredo de um item ou lugar |
