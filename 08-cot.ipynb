## MISSÃO DO DIA

> ❓ **Seu NPC sábio precisa resolver um enigma da dungeon, mas ele sempre dá a resposta direto sem explicar o raciocínio. Como fazer ele "pensar em voz alta" como um detetive?**

Hoje você aprende **Chain of Thought (CoT)** — forçar a IA a mostrar o passo a passo do raciocínio antes de dar a resposta final.

---

## 🗺️ O Mapa desta Aula

| # | O que vamos fazer |
|---|---|
| 1 | Sem CoT: NPC responde direto |
| 2 | Com CoT: NPC raciocina em voz alta |
| 3 | CoT em decisão estratégica de batalha |
| 4 | Seu turno: NPC detective |

## 📖 O Conceito

**Chain of Thought** é como pedir ao mago do grupo que explique o raciocínio antes de lançar o feitiço:

Sem CoT → *"Ataque o dragão pelo flanco esquerdo."*

Com CoT → *"Deixa eu pensar... o dragão tem a asa direita machucada, então o flanco esquerdo está mais exposto. Além disso, o vento sopra para leste, o que favorece chamas da esquerda para direita. Logo — ataque o flanco esquerdo."*

O raciocínio visível ajuda em tarefas que exigem análise, comparação ou decisões.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Sem CoT vs. Com CoT

```python
enigma = """
Na sala há três alavancas: Vermelha, Azul e Verde.
A inscrição diz: 'A verdade fica entre duas mentiras.'
Apenas uma alavanca abre a porta. Qual é a correta?
"""

# Sem CoT: resposta direta
sem_cot = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um mago sábio. Responda qual alavanca puxar."},
        {"role": "user",   "content": enigma}
    ],
    temperature=0.3
)
print("=== SEM CoT ===")
print(sem_cot.choices[0].message.content)

# Com CoT: raciocínio explícito dentro de <pensamento>
com_cot = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é Merlindor, o mago. Sempre mostre seu raciocínio dentro de <pensamento> antes da resposta final."},
        {"role": "user",   "content": enigma}
    ],
    temperature=0.3
)
print("\n=== COM CoT ===")
print(com_cot.choices[0].message.content)
```

## 🔍 Demonstração 2 — CoT em Decisão de Batalha

```python
system_estrategista = """
Você é o General Kavros, estrategista de guerra.
Ao analisar batalhas, use este processo:

<analise>
1. Avalie as forças do inimigo
2. Identifique os pontos fracos
3. Calcule os riscos para o grupo
</analise>

Depois da análise, apresente um PLANO DE 3 MOVIMENTOS.
"""

batalha = """
O grupo enfrenta: 1 Troll (alto HP, regeneração), 3 Goblin Arqueiros (frágeis, à distância).
Nosso grupo: 1 guerreiro (força alta), 1 maga (magia de fogo), 1 ladino (furtividade).
"""

r = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_estrategista},
        {"role": "user",   "content": batalha}
    ],
    temperature=0.4
)
print(r.choices[0].message.content)
```

## 💡 O que aconteceu?

Com CoT, o NPC expõe o raciocínio — isso é útil quando:
- A tarefa envolve análise e comparação
- Você quer entender por que a IA chegou àquela conclusão
- O resultado precisa ser justificado

> ⚠️ **Cuidado:** CoT gera respostas mais longas. Para perguntas simples, não precisa — use só quando a tarefa exige raciocínio.

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Chain of Thought (CoT) | Raciocínio explícito passo a passo antes da resposta | Mago pensando em voz alta antes de lançar feitiço |
| `<pensamento>` | Tag para separar raciocínio da resposta final | O monólogo interno do NPC |
| Transparência | Expor o processo de decisão | O general explicando a estratégia antes da batalha |
