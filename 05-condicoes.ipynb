# 🎮 Aula 05 — Condicionais: NPCs que Reagem ao Contexto

---

## MISSÃO DO DIA

> ❓ **O guarda da cidade deveria ser hostil com ladrões e respeitoso com heróis famosos. Como você faz o NPC mudar de comportamento dependendo de quem está na frente dele?**

Hoje você aprende a usar **condicionais no system prompt** — as regras que dizem ao NPC como reagir dependendo do contexto.

---

## 🗺️ O Mapa desta Aula

| # | O que vamos fazer |
|---|---|
| 1 | NPC que adapta o tom pela classe do jogador |
| 2 | NPC que reage à reputação |
| 3 | NPC com múltiplas condições |
| 4 | Fallback: o que o NPC faz quando não entende |

## 📖 O Conceito

Condicionais no prompt são como as **regras de comportamento do NPC no manual do mestre**:

```
SE o jogador é um nobre: trate com respeito formal
SE o jogador é um ladrão: desconfie e vigie o inventário
SE não sabe quem é: peça documentos
```

Você escreve essas regras no `system` e a IA as segue automaticamente.

## ⚙️ Setup — rode esta célula primeiro

````python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
````

## 🔍 Demonstração 1 — NPC que Adapta o Tom pela Classe

````python
system_guarda = """
Você é o Guarda Torvin, da entrada da cidade de Marveth.
Adapte seu comportamento pela classe do visitante:

SE a classe for GUERREIRO: seja respeitoso e fale de batalhas
SE a classe for MAGO: seja cauteloso e peça registro de feitiços
SE a classe for LADINO: seja desconfiado, mão na espada
SE a classe for CURANDEIRO: seja acolhedor, a cidade precisa de cura
"""

for classe in ["Guerreiro", "Mago", "Ladino"]:
    r = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": system_guarda},
            {"role": "user",   "content": f"Sou um {classe}. Quero entrar na cidade."}
        ],
        temperature=0.6, max_completion_tokens=120
    )
    print(f"\n--- {classe} ---")
    print(r.choices[0].message.content)
````

## 🔍 Demonstração 2 — NPC que Reage à Reputação

````python
system_estalajadeiro = """
Você é Marta, a estalajadeira do Porco Dourado.

SE a reputação for HERÓI LENDÁRIO (>100 pontos): ofereça o quarto VIP de graça
SE a reputação for AVENTUREIRO (50-100 pontos): preço normal e boa conversa
SE a reputação for DESCONHECIDO (<50 pontos): preço dobrado e pagamento adiantado
SE a reputação for CRIMINOSO PROCURADO: diga que não tem quartos
"""

reputacoes = [
    "Sou o Herói Lendário Aldric, tenho 150 pontos de reputação.",
    "Sou um aventureiro com 70 pontos de reputação.",
    "Sou um criminoso procurado."
]

for rep in reputacoes:
    r = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": system_estalajadeiro},
            {"role": "user",   "content": rep + " Quero um quarto."}
        ],
        temperature=0.6, max_completion_tokens=100
    )
    print(f"\n--- Jogador: {rep[:50]}... ---")
    print(r.choices[0].message.content)
````

## 🔍 Demonstração 3 — Fallback: o NPC não entende

````python
# O fallback é o que o NPC faz quando a situação não se encaixa em nenhuma regra
system_oraculo = """
Você é o Oráculo Misterioso. Responda perguntas sobre:

SE for sobre DESTINO: faça uma profecia enigmática
SE for sobre ITENS PERDIDOS: descreva onde encontrar
SE for sobre INIMIGOS: revele o ponto fraco

CASO CONTRÁRIO (fallback):
- Diga que essa questão está além da sua visão
- Sugira 3 perguntas que você poderia responder
"""

r = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_oraculo},
        {"role": "user",   "content": "Qual é a melhor receita de bolo de cenoura?"}
    ],
    temperature=0.7
)
print(r.choices[0].message.content)
````

Crie um NPC com pelo menos **3 condições diferentes** de comportamento.

````python
# SEU TURNO — crie um NPC com comportamento condicional
system_meu_npc = """
Você é ???. # Defina seu NPC

SE ???: ???  # Condição 1
SE ???: ???  # Condição 2
SE ???: ???  # Condição 3
CASO CONTRÁRIO: ???  # Fallback
"""

situacao = "???"  # Teste uma das condições

r = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_meu_npc},
        {"role": "user",   "content": situacao}
    ],
    temperature=0.7
)
print(r.choices[0].message.content)
````

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Condicional | Regra do tipo SE...ENTÃO no prompt | Comportamento do NPC baseado em contexto |
| Fallback | O que acontece quando nenhuma condição bate | "Não sei responder isso" do NPC |
| Contexto | As informações que o jogador passa ao NPC | A identidade e situação do personagem |
