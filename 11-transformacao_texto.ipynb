## 📖 O Conceito

**Transformação de texto** é como o **tradutor do reino** que pega o mesmo documento e o adapta para diferentes audiências:

| Transformação | O que muda | Exemplo no RPG |
|---|---|---|
| Formato | Estrutura do texto | Quest → checklist de missão |
| Tom | Estilo e registro | Sério → cômico / formal → casual |
| Perspectiva | Ponto de vista | Narrador → personagem principal |
| Público | Linguagem e complexidade | Adulto → infantil / gamer → não-gamer |

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Transformação de Formato

```python
quest_original = """
O herói deve infiltrar o castelo Sombraverde, localizar o pergaminho roubado no arquivo secreto
do terceiro andar, neutralizar os dois guardas sem fazer barulho, e sair pelo túnel dos fundos
antes do nascer do sol.
"""

# Transformar em checklist
checklist = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Transforme o texto em um checklist de missão. Use:\n☐ [ação]"
},
        {"role": "user",   "content": quest_original}
    ],
    temperature=0.3
)
print("=== QUEST ORIGINAL ===")
print(quest_original)
print("\n=== CHECKLIST ===")
print(checklist.choices[0].message.content)
```

## 🔍 Demonstração 2 — Transformação de Tom

```python
cena_seria = "O dragão Vorathex emergiu das chamas, seus olhos ardendo com ódio milenar. O herói empunhou a espada com mãos trêmulas."

# Tom cômico
comico = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Reescreva o texto com tom completamente cômico e levinho, como se fosse uma cena de comédia."},
        {"role": "user",   "content": cena_seria}
    ],
    temperature=0.9
)

# Post de redes sociais
post = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Transforme em um post curto para redes sociais (máx 2 frases), com emojis e hashtags."},
        {"role": "user",   "content": cena_seria}
    ],
    temperature=0.8
)

print("=== TOM CÔMICO ===")
print(comico.choices[0].message.content)
print("\n=== POST DE REDES SOCIAIS ===")
print(post.choices[0].message.content)
```

## 🔍 Demonstração 3 — Adaptando para Públicos Diferentes

```python
descricao_dungeon = "A dungeon contém criaturas com alto DPS, mecânicas de enrage e um boss com fase de transição ao atingir 30% de HP."

# Para crianças
infantil = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Reescreva para crianças de 8 anos. Linguagem simples, sem jargões técnicos."},
        {"role": "user",   "content": descricao_dungeon}
    ],
    temperature=0.6
)

# Para quem não joga videogame
nao_gamer = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Reescreva para alguém que nunca jogou videogame. Use analogias do cotidiano."},
        {"role": "user",   "content": descricao_dungeon}
    ],
    temperature=0.6
)

print("=== PARA CRIANÇAS ===")
print(infantil.choices[0].message.content)
print("\n=== PARA QUEM NÃO JOGA ===")
print(nao_gamer.choices[0].message.content)
```

```python
# SEU TURNO — defina o texto e as transformações
MEU_TEXTO = "???"  # Qualquer texto do seu jogo

TRANSFORMACAO_1 = "???"  # Ex: 'Reescreva como post de Twitter/X'
TRANSFORMACAO_2 = "???"  # Ex: 'Adapte para crianças de 10 anos'

for transf in [TRANSFORMACAO_1, TRANSFORMACAO_2]:
    r = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": transf},
            {"role": "user",   "content": MEU_TEXTO}
        ],
        temperature=0.7
    )
    print(f"\n=== {transf.upper()} ===")
    print(r.choices[0].message.content)
```

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Transformação de formato | Mudar a estrutura (parágrafo → lista, quest → checklist) | Tradutor do mapa para diferentes formas |
| Transformação de tom | Mudar o registro (sério/cômico, formal/casual) | O mesmo NPC em versão séria e em versão zueira |
| Adaptação de público | Ajustar linguagem para a audiência-alvo | A mesma lenda contada para crianças e adultos |
