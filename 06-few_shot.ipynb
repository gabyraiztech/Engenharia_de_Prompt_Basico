## MISSÃO DO DIA

> ❓ **Você quer que o ferreiro do jogo fale como um velho mal-humorado com sotaque nordestino, sem precisar explicar 10 regras. Como você faz isso?**

Hoje você aprende **few-shot prompting** — ensinar a IA por exemplos, como mostrar ao ator como é o personagem antes de ele entrar em cena.

---

## 🗺️ O Mapa desta Aula

| # | O que vamos fazer |
|---|---|
| 1 | Zero-shot: sem exemplos |
| 2 | One-shot: com um exemplo |
| 3 | Few-shot: com vários exemplos |
| 4 | Few-shot criativo: estilo de personagem |

## 📖 O Conceito

| Técnica | O que é | Analogia do RPG |
|---|---|---|
| **Zero-shot** | Nenhum exemplo, só instrução | Dizer ao ator: "seja engraçado" sem mostrar nada |
| **One-shot** | Um exemplo | Mostrar um monólogo e dizer: "assim" |
| **Few-shot** | Vários exemplos | Ensaio completo com 3-5 cenas de referência |

Quanto mais exemplos consistentes você der, mais o NPC vai manter o estilo certo.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração — Zero, One e Few-Shot lado a lado

```python
tarefa_user = "O herói pergunta: você vende espadas?"

# Zero-shot: só instrução, sem exemplo
zero = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um ferreiro mal-humorado."},
        {"role": "user",   "content": tarefa_user}
    ],
    temperature=0.5
)
print("=== ZERO-SHOT ===")
print(zero.choices[0].message.content)

# One-shot: um exemplo de como o ferreiro fala
one = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system",    "content": "Você é Brun, o ferreiro."},
        {"role": "user",      "content": "Você faz escudos?"},
        {"role": "assistant", "content": "Escudo?! Claro que faço, rapaiz! Mas não é pra qualquer zé-mané não, não!"},
        {"role": "user",      "content": tarefa_user}
    ],
    temperature=0.5
)
print("\n=== ONE-SHOT ===")
print(one.choices[0].message.content)

# Few-shot: múltiplos exemplos para fixar o estilo
few = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system",    "content": "Você é Brun, o ferreiro. Fale sempre com linguagem simples, sotaque nordestino e muito orgulho do seu trabalho."},
        {"role": "user",      "content": "Você faz escudos?"},
        {"role": "assistant", "content": "Escudo?! Claro que faço, rapaiz! Mas não é pra qualquer zé-mané não, não!"},
        {"role": "user",      "content": "Quanto tempo leva?"},
        {"role": "assistant", "content": "Depende, visse? Armadura completa leva uns três dias. Mas eu não faço trabalho xexelento, não — cada peça é obra de arte!"},
        {"role": "user",      "content": "Aceita troca por poções?"},
        {"role": "assistant", "content": "Poção?! Esse negócio de magia não me atrai, não. Prefiro ouro bom e honesto, oxente!"},
        {"role": "user",      "content": tarefa_user}
    ],
    temperature=0.5
)
print("\n=== FEW-SHOT ===")
print(few.choices[0].message.content)
```

## 💡 O que aconteceu?

Com **zero-shot**, o ferreiro é genérico — qualquer ferreiro de qualquer jogo.

Com **few-shot**, o Brun tem **voz própria**: sotaque, trejeitos, personalidade consistente. Isso é o que faz um NPC memorável.

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Zero-shot | Sem exemplos — só instrução | Dizer ao ator o papel sem ensaio |
| One-shot | Um exemplo | Uma cena de referência |
| Few-shot | Vários exemplos (3-5) | Ensaio completo com cenas-modelo |
| Estilo | Voz e personalidade consistente do NPC | A "alma" do personagem |
