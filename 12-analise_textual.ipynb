## 📖 O Conceito

**Análise textual** é usar a IA como um **crítico literário** — ela lê o que você escreveu e devolve uma análise profunda.

| Tipo de análise | O que revela | No RPG é... |
|---|---|---|
| Tom e atmosfera | Sentimento que o texto transmite | Se a dungeon realmente parece assustadora |
| Símbolos | Elementos com significado oculto | O dragão como símbolo de corrupção |
| Consistência | Contradições na narrativa | "O mago disse que não pode voar, mas depois voou" |
| Comparativa | Diferenças entre dois textos | O antes e depois de uma cena reescrita |

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Análise de Tom e Atmosfera

```python
cena_dungeon = """
A porta rangeu ao abrir. Dentro, o ar estava pesado, com cheiro de pedra úmida e algo mais antigo —
como um silêncio que tinha peso. As tochas nas paredes tremiam sem motivo aparente.
No centro da sala, um altar de obsidiana com manchas que poderiam ser ferrugem... ou não.
"""

analise_tom = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um crítico literário especializado em narrativas de fantasia. Analise:\n1. Tom e atmosfera transmitidos\n2. Recursos literários usados\n3. Impacto emocional no leitor"},
        {"role": "user",   "content": cena_dungeon}
    ],
    temperature=0.4
)
print("=== ANÁLISE DE TOM ===")
print(analise_tom.choices[0].message.content)
```

## 🔍 Demonstração 2 — Análise Simbólica do Lore

```python
lore_mundo = """
No princípio, havia apenas a Chama Eterna e o Vazio. A Chama criou os mortais do calor e da luz.
Mas o Vazio sussurrou nos sonhos dos mortais, prometendo poder sem limite.
Os que cederam ao Vazio tornaram-se os Sombrios — nem vivos, nem mortos.
Dizem que toda vez que um mortal escolhe o poder sobre o amor, uma nova chama se apaga no mundo.
"""

simbolos = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um semioticista de mundos de fantasia. Identifique:\n1. Símbolos principais e seus significados\n2. Metáforas presentes\n3. Temas universais abordados"},
        {"role": "user",   "content": lore_mundo}
    ],
    temperature=0.4
)
print("=== ANÁLISE SIMBÓLICA ===")
print(simbolos.choices[0].message.content)
```

## 🔍 Demonstração 3 — Verificando Consistência

```python
lore_com_falha = """
O reino de Avareth nunca foi conquistado em 500 anos de história.
O castelo é impenetrável — paredes de granito de 10 metros de espessura.
Nenhum inimigo jamais entrou na capital.
... [sessão 3] ...
O herói descobriu que o Império havia ocupado Avareth há 200 anos durante a Grande Guerra.
Os registros da ocupação ainda estão no arquivo da capital.
"""

consistencia = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Você é um editor de lore. Identifique contradições, inconsistências e buracos na narrativa. Seja específico."},
        {"role": "user",   "content": lore_com_falha}
    ],
    temperature=0.3
)
print("=== ANÁLISE DE CONSISTÊNCIA ===")
print(consistencia.choices[0].message.content)
```
