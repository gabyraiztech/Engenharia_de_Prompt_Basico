## 📖 O Conceito

Escrever um bom prompt é como entregar a **ficha de quest** para um narrador de RPG.

| Situação | Resultado |
|---|---|
| "Crie uma quest" | O narrador inventa algo genérico — um dragão que rouba aldeias |
| "Crie uma quest de 3 etapas para um ladino nível 5, passando por uma taverna, uma floresta e um castelo" | O narrador cria exatamente o que você precisava |

**As 4 regras de ouro:**
1. Use **verbos de ação** (crie, liste, explique, compare)
2. Seja **específico** (personagem, nível, ambiente, número de etapas)
3. **Controle o formato** (tabela, lista, parágrafo)
4. Use **separadores** quando misturar contexto e instrução

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração 1 — Prompt Vago vs. Específico

```python
# Prompt vago — o narrador vai na direção que quiser
vago = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":"Crie uma quest de RPG"}],
    temperature=0.7, max_completion_tokens=200
)
print("=== PROMPT VAGO ===")
print(vago.choices[0].message.content)

# Prompt específico — o narrador tem um roteiro claro
especifico = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":
        "Crie uma quest de RPG medieval com EXATAMENTE 3 etapas.\n"
        "Personagem: ladino nível 5, especialidade em furtividade.\n"
        "Ambientes: taverna → floresta → castelo.\n"
        "Tom: suspense e mistério. Sem dragões."
    }],
    temperature=0.7, max_completion_tokens=300
)
print("\n=== PROMPT ESPECÍFICO ===")
print(especifico.choices[0].message.content)
```

## 💡 O que aconteceu?

O prompt específico entregou uma quest com estrutura, personagem e ambientes definidos — porque você deu todas as peças ao narrador.

O prompt vago deu ao narrador liberdade total — que pode ser boa ou péssima, dependendo do que você precisa.

## 🔍 Demonstração 2 — Controlando o Formato

```python
# Sem controle de formato — o narrador decide como apresentar
sem_formato = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":"Liste recompensas de dungeon"}],
    temperature=0.5, max_completion_tokens=150
)
print("=== SEM FORMATO ===")
print(sem_formato.choices[0].message.content)

# Com formato definido — saída organizada e usável
com_formato = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":
        "Liste exatamente 4 recompensas de dungeon.\n"
        "Use o formato: - [Nome do item]: [efeito em 1 frase]"
    }],
    temperature=0.5, max_completion_tokens=150
)
print("\n=== COM FORMATO ===")
print(com_formato.choices[0].message.content)
```

## 🔍 Demonstração 3 — Usando Separadores

```python
# Separadores deixam claro o que é contexto e o que é instrução
com_separador = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":
        """Analise a lore abaixo e crie 2 ganchos de missão.

--- LORE DO MUNDO ---
O reino de Avareth foi destruído há 300 anos por uma praga mágica.
Apenas uma torre sobreviveu: a Torre dos Ecos.
Dizem que vozes do passado ainda ecoam lá dentro.
--- FIM DA LORE ---

TAREFA: Crie 2 ganchos de missão inspirados nesta lore. Cada gancho em 1 frase."""
    }],
    temperature=0.7
)
print(com_separador.choices[0].message.content)
```

## 💡 O que aconteceu?

Os separadores `--- LORE ---` deixam claro o que é o contexto dado por você e o que é a tarefa. Sem isso, a IA pode confundir o que deve ler com o que deve fazer.

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Prompt específico | Instrução com contexto, formato e restrições claras | Quest bem escrita com objetivos claros |
| Prompt vago | Instrução aberta sem direcionamento | "Vá lá e faça alguma coisa" |
| Separador | Marcador que divide contexto de instrução (`---`) | Fronteira entre o mapa e as instruções do mestre |
| Formato controlado | Especificar como a resposta deve ser estruturada | Ficha de personagem com campos definidos |
