## 🔍 Demonstração — [NOME DO EXEMPLO]

[1-2 frases explicando O QUE o código abaixo vai mostrar, dentro do contexto do RPG]

```python
# [Comentário único descrevendo o bloco, não cada linha]
resposta = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "[SYSTEM PROMPT RELACIONADO AO RPG]"},
        {"role": "user",   "content": "[PERGUNTA/AÇÃO DO JOGADOR]"}
    ],
    temperature=0.7
)

print(resposta.choices[0].message.content)
```
