## 📖 O Conceito

**Multi-step** é como um **mestre de jogo experiente** que não improvisa tudo de uma vez:

1. Define o **tema** da dungeon
2. Cria os **andares e ambientes**
3. Popula com **monstros e armadilhas**
4. Esconde os **tesouros**
5. Cria o **chefe final**

Cada etapa usa o resultado da anterior — a dungeon cresce de forma coesa e controlada.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração — Dungeon em 5 Passos

```python
system_dungeon = """
Você é um arquiteto de dungeons mestre. Siga a metodologia de 5 passos:
PASSO 1: Defina o tema e nome da dungeon
PASSO 2: Descreva os 3 andares com ambiente único em cada um
PASSO 3: Popule cada andar com 1 inimigo e 1 armadilha
PASSO 4: Esconda 1 tesouro especial por andar
PASSO 5: Crie o chefe final com pontos fracos e forte
Apresente cada passo com título claro.
"""

dungeon = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": system_dungeon},
        {"role": "user",   "content": "Crie uma dungeon de tema aquático para um grupo de nível 4."}
    ],
    temperature=0.6
)
print(dungeon.choices[0].message.content)
```

## 🔍 Comparação — Single-Step vs. Multi-Step

```python
# Single-step: tudo de uma vez
single = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role":"user","content":"Crie uma dungeon de RPG"}],
    temperature=0.6, max_completion_tokens=200
)
print("=== SINGLE-STEP ===")
print(single.choices[0].message.content)

# Multi-step: estruturado e metodológico
multi = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
        {"role": "system", "content": "Siga 3 passos:\nPASSO 1: Nome e tema\nPASSO 2: 2 ambientes\nPASSO 3: 1 chefe com ponto fraco"},
        {"role": "user",   "content": "Crie uma dungeon de vulcão para nível 8"}
    ],
    temperature=0.6, max_completion_tokens=300
)
print("\n=== MULTI-STEP ===")
print(multi.choices[0].message.content)
```

## 💡 O que aconteceu?

O single-step inventa tudo de uma vez — pode ser criativo, mas sem estrutura.

O multi-step **garante que cada componente existe** porque o modelo é obrigado a passar por cada etapa. Ideal quando o resultado precisa ser completo e coerente.

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| Multi-step | Dividir uma tarefa em etapas sequenciais | Planejar a dungeon andar por andar |
| Single-step | Tudo em um único prompt | Improvisar a dungeon na hora |
| Metodologia | Conjunto de passos definidos no system | As regras do mestre para criar aventuras |
