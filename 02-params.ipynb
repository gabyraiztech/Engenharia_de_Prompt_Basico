## 📖 O Conceito

**Temperature** é o "botão de criatividade" do NPC.

| Temperature | O NPC é como... | Quando usar |
|---|---|---|
| `0.0 – 0.3` | Guarda da porta: sempre diz a mesma frase | Respostas técnicas, fichas, regras do jogo |
| `0.4 – 0.7` | Bardo equilibrado: varia mas faz sentido | Diálogos gerais, NPCs normais |
| `0.8 – 1.0` | Oráculo inspirado: poético e imprevisível | Vilões, profecias, NPCs excêntricos |
| `1.0+` | Mago louco: pode não fazer sentido | Apenas para efeitos de comédia ou terror |

**top_p** controla o "vocabulário que a IA considera" — valores baixos = palavras mais prováveis, valores altos = mais variedade.

## ⚙️ Setup — rode esta célula primeiro

```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.environ.get("GROQ_API"))
```

## 🔍 Demonstração — O Mesmo NPC em 3 Humores

O oráculo **Vel'Zar** vai responder a mesma pergunta com temperaturas diferentes. Veja como a personalidade muda.

```python
pergunta = "O que o destino reserva para mim, oráculo?"

for temp, humor in [(0.0, "Frio/Robótico"), (0.7, "Equilibrado"), (1.5, "Loucura Total")]:
    resposta = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": "Você é Vel'Zar, um oráculo antigo de fantasia medieval."},
            {"role": "user",   "content": pergunta}
        ],
        temperature=temp
    )
    print(f"\n--- Temperature {temp} ({humor}) ---")
    print(resposta.choices[0].message.content)
```

## 💡 O que aconteceu?

Com **temperature 0.0** o oráculo é previsível — bom para respostas que precisam ser consistentes (como regras do jogo).

Com **temperature 0.7** ele tem personalidade sem perder coerência — ideal para NPCs de diálogo.

Com **temperature 1.5** ele pode surpreender (ou não fazer sentido) — cuidado, use só em NPCs propositalmente loucos.

## 🔍 Demonstração 2 — top_p em ação

```python
# top_p baixo = NPC usa palavras mais comuns e previsíveis
# top_p alto  = NPC tem vocabulário mais variado e criativo

for top_p_val, estilo in [(0.1, "Direto"), (0.9, "Poético")]:
    resposta = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": "Você é um narrador de RPG medieval."},
            {"role": "user",   "content": "Descreva a entrada de uma dungeon escura."}
        ],
        temperature=1.0,
        top_p=top_p_val
    )
    print(f"\n--- top_p {top_p_val} ({estilo}) ---")
    print(resposta.choices[0].message.content)
```

## 💡 O que aconteceu?

Com **top_p 0.1** o narrador escolhe as palavras mais óbvias — claro e direto.

Com **top_p 0.9** ele considera uma gama maior de palavras — mais rico e variado.

> 💡 Na prática: use **temperature** para controlar criatividade e **top_p** para controlar variedade de vocabulário. Raramente você precisa dos dois ao mesmo tempo.

## 📚 Glossário da Aula

| Termo | Definição | No RPG é... |
|---|---|---|
| `temperature` | Controla aleatoriedade da resposta (0.0 a 2.0) | O humor do NPC: calmo ↔ louco |
| `top_p` | Controla diversidade do vocabulário (0.0 a 1.0) | O vocabulário que o NPC considera usar |
| Determinístico | Sempre a mesma resposta | NPC com script fixo |
| Estocástico | Resposta varia a cada vez | NPC com diálogos aleatórios |
